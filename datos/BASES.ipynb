#### ============================================================ 
#### MARKETING – CLASIFICACIÓN
#### ============================================================

TIPO DE PROBLEMA Clasificación.

PREGUNTA DE INVESTIGACIÓN ¿Qué características de los clientes y de las campañas de marketing permiten predecir si un cliente contratará un depósito a lazo?

OBJETIVO Predecir si un cliente contratará un depósito a plazo luego de una campaña de marketing bancario.

VARIABLE OBJETIVO (y) y

Valores: - yes: el cliente contrató el depósito. - no: el cliente no contrató el depósito.


VARIABLES PREDICTORAS (X) Las restantes variables del dataset, entre
ellas:

-   age: edad del cliente.
-   job: tipo de trabajo.
-   education: nivel educativo.
-   balance: saldo promedio anual.
-   housing: indica si posee préstamo hipotecario.
-   loan: indica si posee préstamo personal.
-   duration: duración del último contacto.
-   campaign: cantidad de contactos realizados durante la campaña.
-   poutcome: resultado de la campaña de marketing anterior.


#### ============================================================ 
#### NEWS POPULARITY – REGRESIÓN
#### ============================================================

TIPO DE PROBLEMA Regresión.

PREGUNTA DE INVESTIGACIÓN: ¿Qué características de una noticia permiten predecir la cantidad de veces que será compartida?


OBJETIVO Predecir la popularidad de una noticia, utilizando como medida
la cantidad de veces que fue compartida.

VARIABLE OBJETIVO (y) shares

VARIABLES PREDICTORAS (X) 
-   n_tokens_title: cantidad de palabras del título.
-   n_tokens_content: cantidad de palabras del contenido.
-   num_self_hrefs: cantidad de enlaces internos.
-   num_imgs: cantidad de imágenes.
-   num_videos: cantidad de videos.
-   num_keywords: cantidad de palabras clave.
-   global_subjectivity: nivel global de subjetividad.
-   global_sentiment_polarity: polaridad global del sentimiento.
-   title_subjectivity: subjetividad del título.
-   title_sentiment_polarity: sentimiento del título.
